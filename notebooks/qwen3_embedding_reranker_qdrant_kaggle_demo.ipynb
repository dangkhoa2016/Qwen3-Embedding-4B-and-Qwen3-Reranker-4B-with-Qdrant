## 2. Clean up a verified prior notebook run / Dọn dẹp lần chạy notebook trước đã được xác minh

**English:** This makes repeated **Run All** in the same Kaggle session safe. The notebook never terminates a process merely because a port is busy. It first proves ownership using PID files and `/proc` evidence: the FastAPI process must be the repository's `uvicorn qwen3_qdrant.main:app` on port `8000` with the repository as its working directory; the Qwen3-Reranker-4B llama.cpp process must use the exact hardened launcher SHA-256, the expected GGUF and port `8081`; and Qdrant must be the expected `1.18.3` binary with the notebook-owned storage path. Only verified-owned processes receive `SIGTERM`, with `SIGKILL` used only if the same PID/start-time is still active after the grace period. After verified cleanup, only notebook-owned workspaces are reset. If any of ports `6333`, `8000`, or `8081` remains occupied by an unverified process, the notebook fails closed rather than killing it.

**Tiếng Việt:** Phần này giúp **Run All** lặp lại trong cùng Kaggle session một cách an toàn. Notebook không bao giờ terminate process chỉ vì một port đang bận. Trước tiên notebook chứng minh ownership bằng PID files và `/proc`: FastAPI process phải là `uvicorn qwen3_qdrant.main:app` của repository trên port `8000` với working directory là repository; llama.cpp process của Qwen3-Reranker-4B phải dùng đúng hardened launcher SHA-256, đúng GGUF và port `8081`; Qdrant phải là binary `1.18.3` dự kiến với notebook-owned storage path. Chỉ process đã verify ownership mới nhận `SIGTERM`; `SIGKILL` chỉ được dùng nếu cùng PID/start-time vẫn active sau grace period. Sau khi cleanup đã verify, notebook chỉ reset các workspace do chính notebook sở hữu. Nếu port `6333`, `8000` hoặc `8081` vẫn bị một process chưa xác minh chiếm, notebook fail closed thay vì kill process đó.


````python
from pathlib import Path
import hashlib
import json
import os
import shutil
import signal
import socket
import time

print("PRIOR_RUN_CLEANUP_BEGIN")

PROC_ROOT = Path("/proc")
OWNED_RUN_ROOT = Path("/kaggle/working/qwen3-hybrid-qdrant-production-demo")
OWNED_RUNTIME_ROOT = Path("/kaggle/working/qwen3-reranker-4b-hardened-llama-cpp-runtime")
EXPECTED_QDRANT_EXE = Path("/kaggle/working/qdrant-1.18.3/bin/qdrant")
EXPECTED_QDRANT_STORAGE = OWNED_RUN_ROOT / "qdrant/storage"
EXPECTED_LAUNCHER_SHA = "28a79707376877f09065fa05fda5a9a6f57dfb4aed01c9918123667e38ae1f41"
EXPECTED_IMPL_SHA = "c4807f2f10cdf354270ac97c1f091d0846e8154749b4b8347f5f26a40184d425"
EXPECTED_GGUF_NAME = "Qwen3-Reranker-4B.Q4_K_M.gguf"

def _proc_dir(pid):
    return PROC_ROOT / str(int(pid))

def _proc_state(pid):
    try:
        raw = (_proc_dir(pid) / "stat").read_text()
        tail = raw[raw.rfind(")") + 2:].split()
        return tail[0]
    except (OSError, ValueError, IndexError):
        return None

def _proc_starttime(pid):
    try:
        raw = (_proc_dir(pid) / "stat").read_text()
        tail = raw[raw.rfind(")") + 2:].split()
        # tail[0] is field 3 (state), so tail[19] is field 22 (starttime).
        return int(tail[19])
    except (OSError, ValueError, IndexError):
        return None

def _same_process_starttime(pid, starttime):
    return starttime is not None and _proc_starttime(pid) == starttime

def _proc_active(pid):
    return _proc_dir(pid).is_dir() and _proc_state(pid) not in {None, "Z"}

def _proc_args(pid):
    try:
        return [
            part.decode("utf-8", "replace")
            for part in (_proc_dir(pid) / "cmdline").read_bytes().split(b"\0")
            if part
        ]
    except (OSError, PermissionError):
        return []

def _proc_cwd(pid):
    try:
        return Path(os.readlink(_proc_dir(pid) / "cwd")).resolve()
    except (OSError, PermissionError):
        return None

def _proc_exe_link(pid):
    try:
        return os.readlink(_proc_dir(pid) / "exe")
    except (OSError, PermissionError):
        return ""

def _proc_exe_sha(pid):
    try:
        with open(_proc_dir(pid) / "exe", "rb") as handle:
            return hashlib.sha256(handle.read()).hexdigest()
    except (OSError, PermissionError):
        return None

def _proc_environ(pid):
    try:
        parts = (_proc_dir(pid) / "environ").read_bytes().split(b"\0")
    except (OSError, PermissionError):
        return {}
    out = {}
    for part in parts:
        if not part or b"=" not in part:
            continue
        key, value = part.split(b"=", 1)
        out[key.decode("utf-8", "replace")] = value.decode("utf-8", "replace")
    return out

def _has_cli_port(args, port):
    target = str(port)
    for i, arg in enumerate(args):
        if arg == f"--port={target}":
            return True
        if arg == "--port" and i + 1 < len(args) and args[i + 1] == target:
            return True
    return False

def _is_owned_hybrid(pid):
    args = _proc_args(pid)
    joined = " ".join(args)
    return (
        _proc_active(pid)
        and _proc_cwd(pid) == APP_ROOT.resolve()
        and "uvicorn" in joined
        and "qwen3_qdrant.main:app" in joined
        and _has_cli_port(args, 8000)
    )

def _is_owned_llama(pid):
    args = _proc_args(pid)
    joined = " ".join(args)
    if not (
        _proc_active(pid)
        and _proc_cwd(pid) == APP_ROOT.resolve()
        and EXPECTED_GGUF_NAME in joined
        and _has_cli_port(args, 8081)
        and _proc_exe_sha(pid) == EXPECTED_LAUNCHER_SHA
    ):
        return False

    maps_path = _proc_dir(pid) / "maps"
    try:
        maps_text = maps_path.read_text(errors="replace")
    except (OSError, PermissionError):
        return False

    # Require evidence that the hardened implementation is mapped. If the
    # previous run's files were already unlinked, the exact launcher identity,
    # working directory, GGUF, and port still remain mandatory ownership gates.
    impl_mapped = "libllama-server-impl.so" in maps_text
    return impl_mapped

def _is_owned_qdrant(pid):
    if not _proc_active(pid):
        return False
    exe_link = _proc_exe_link(pid).removesuffix(" (deleted)")
    try:
        exe_ok = Path(exe_link).resolve() == EXPECTED_QDRANT_EXE.resolve()
    except OSError:
        exe_ok = False
    env = _proc_environ(pid)
    return (
        exe_ok
        and env.get("QDRANT__STORAGE__STORAGE_PATH") == str(EXPECTED_QDRANT_STORAGE)
        and env.get("QDRANT__SERVICE__HTTP_PORT") == "6333"
        and env.get("QDRANT__SERVICE__HOST") == "127.0.0.1"
    )

def _pid_from_file(path):
    try:
        raw = path.read_text().strip()
        return int(raw) if raw else None
    except (OSError, ValueError):
        return None

def _scan_owned(predicate):
    found = set()
    for proc in PROC_ROOT.iterdir():
        if not proc.name.isdigit():
            continue
        pid = int(proc.name)
        try:
            if predicate(pid):
                found.add(pid)
        except (OSError, PermissionError):
            continue
    return found

def _add_if_verified(candidates, pid, label, predicate):
    if not isinstance(pid, int) or not _proc_active(pid):
        return
    if predicate(pid):
        candidates.add(pid)
    else:
        print(f"PRIOR_RUN_STALE_PID_IGNORED label={label} pid={pid}")

def _wait_inactive(pid, starttime, timeout=12.0):
    deadline = time.monotonic() + timeout
    while time.monotonic() < deadline:
        if not _same_process_starttime(pid, starttime) or not _proc_active(pid):
            return True
        time.sleep(0.1)
    return not _same_process_starttime(pid, starttime) or not _proc_active(pid)

def _terminate_verified(pid, label, predicate):
    if pid is None or not _proc_active(pid):
        return False
    starttime = _proc_starttime(pid)
    if not predicate(pid):
        raise RuntimeError(
            f"{label} pid={pid} is active but ownership verification failed; refusing to terminate it"
        )

    print(f"PRIOR_RUN_OWNED_PROCESS={label} pid={pid} action=SIGTERM")
    os.kill(pid, signal.SIGTERM)
    if _wait_inactive(pid, starttime):
        return True

    if not _same_process_starttime(pid, starttime):
        return True
    if not predicate(pid):
        raise RuntimeError(
            f"{label} pid={pid} changed identity before SIGKILL; refusing to continue"
        )

    print(f"PRIOR_RUN_OWNED_PROCESS={label} pid={pid} action=SIGKILL")
    os.kill(pid, signal.SIGKILL)
    if not _wait_inactive(pid, starttime, timeout=5.0):
        raise RuntimeError(f"{label} pid={pid} did not stop")
    return True

def _port_open(port):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        sock.settimeout(0.25)
        return sock.connect_ex(("127.0.0.1", int(port))) == 0

def _assert_ports_free():
    busy = [port for port in (6333, 8000, 8081) if _port_open(port)]
    if busy:
        print("PRIOR_RUN_UNOWNED_PORT=FAIL ports=", busy)
        raise RuntimeError(
            f"Ports remain occupied after verified-owned cleanup: {busy}. "
            "Refusing to terminate unverified processes."
        )
    print("PRIOR_RUN_PORTS_FREE=PASS")

stopped = 0

# Prefer PID files/ownership record, while also scanning for verified-owned
# processes so a rerun can clean up an earlier notebook revision safely.
owner_record_path = OWNED_RUN_ROOT / "qualification-owner.json"
record = {}
if owner_record_path.is_file():
    try:
        record = json.loads(owner_record_path.read_text())
    except Exception:
        record = {}

hybrid_candidates = set()
for candidate in (
    record.get("hybrid_pid"),
    _pid_from_file(OWNED_RUN_ROOT / "hybrid-server.pid"),
):
    _add_if_verified(hybrid_candidates, candidate, "hybrid-api", _is_owned_hybrid)
hybrid_candidates |= _scan_owned(_is_owned_hybrid)

for pid in sorted(hybrid_candidates):
    stopped += int(_terminate_verified(pid, "hybrid-api", _is_owned_hybrid))

# Graceful FastAPI shutdown should normally stop its Qwen3-Reranker-4B child.
# Any remaining runtime process is independently re-verified before termination.
llama_candidates = set()
for candidate in (
    record.get("llama_pid"),
    _pid_from_file(OWNED_RUN_ROOT / "llama-server.pid"),
):
    _add_if_verified(llama_candidates, candidate, "reranker-runtime", _is_owned_llama)
llama_candidates |= _scan_owned(_is_owned_llama)

for pid in sorted(llama_candidates):
    stopped += int(_terminate_verified(pid, "reranker-runtime", _is_owned_llama))

qdrant_candidates = set()
for candidate in (
    record.get("qdrant_pid"),
    _pid_from_file(OWNED_RUN_ROOT / "qdrant/qdrant.pid"),
):
    _add_if_verified(qdrant_candidates, candidate, "qdrant", _is_owned_qdrant)
qdrant_candidates |= _scan_owned(_is_owned_qdrant)

for pid in sorted(qdrant_candidates):
    stopped += int(_terminate_verified(pid, "qdrant", _is_owned_qdrant))

_assert_ports_free()

# Reset only directories owned by this notebook after all owned services stop.
for owned_path in (OWNED_RUN_ROOT, OWNED_RUNTIME_ROOT):
    if owned_path.exists():
        shutil.rmtree(owned_path)

print("PRIOR_RUN_OWNED_PROCESSES_STOPPED =", stopped)
print("PRIOR_RUN_WORKSPACE_RESET=PASS")
print("PRIOR_RUN_CLEANUP=PASS")
````

## 3. Start the qualification timer and pin the canonical CPU profile / Bắt đầu qualification timer và khóa canonical CPU profile

````python
RUN_ALL_STARTED = time.perf_counter()
RUN_ALL_BUDGET_SECONDS = 600

def read_memory_events():
    path = Path("/sys/fs/cgroup/memory.events")
    out = {}
    if path.exists():
        for line in path.read_text().splitlines():
            key, value = line.split()
            out[key] = int(value)
    return out

MEMORY_EVENTS_BEFORE = read_memory_events()

RETRIEVAL_TOP_K = 5
RERANK_TOP_K = 5
DISPLAY_TOP_K = 5
LLAMA_SERVER_THREADS = 2
TORCH_NUM_THREADS = 2

# Example stronger host only — do not use for the canonical reproduction:
# RETRIEVAL_TOP_K = 50
# RERANK_TOP_K = 50
# DISPLAY_TOP_K = 10
# LLAMA_SERVER_THREADS = 8
# TORCH_NUM_THREADS = 8

assert 1 <= DISPLAY_TOP_K <= RERANK_TOP_K <= RETRIEVAL_TOP_K
assert RERANK_TOP_K <= 200

for key, value in {
    "RETRIEVAL_TOP_K": RETRIEVAL_TOP_K,
    "RERANK_TOP_K": RERANK_TOP_K,
    "DISPLAY_TOP_K": DISPLAY_TOP_K,
    "LLAMA_SERVER_THREADS": LLAMA_SERVER_THREADS,
    "TORCH_NUM_THREADS": TORCH_NUM_THREADS,
    "MAX_RERANK_DOCUMENTS": RERANK_TOP_K,
    "TORCH_NUM_INTEROP_THREADS": 1,
    "MAX_CONCURRENT_INFERENCE": 1,
}.items():
    os.environ[key] = str(value)

print({
    key: os.environ[key]
    for key in [
        "RETRIEVAL_TOP_K",
        "RERANK_TOP_K",
        "DISPLAY_TOP_K",
        "LLAMA_SERVER_THREADS",
        "TORCH_NUM_THREADS",
    ]
})
````

## 4. Install source and stage the qualified hardened llama.cpp runtime / Cài source và stage qualified hardened llama.cpp runtime

**English:** Install the checked-out source, prepare Qdrant, then locate the attached `Qwen3-Reranker-4B` hardened llama.cpp runtime under Kaggle's dataset namespace. The canonical path is `/kaggle/input/datasets/dangkhoa2016/qwen3-reranker-4b-hardened-llama-cpp-runtime`; a fail-closed recursive lookup inside `/kaggle/input/datasets` is used only if that exact path is unavailable. The cell verifies all eight entries in `RUNTIME-MANIFEST.sha256`, copies the bundle byte-identically to `/kaggle/working`, sets `LD_LIBRARY_PATH` and `LLAMA_SERVER_BIN`, verifies launcher linkage, and requires `LLAMA_CPP_SETUP=PASS_EXISTING_BINARY`. **Tiếng Việt:** Cài source đã checkout, chuẩn bị Qdrant, sau đó locate hardened llama.cpp runtime đã attach cho `Qwen3-Reranker-4B` bên trong Kaggle dataset namespace. Canonical path là `/kaggle/input/datasets/dangkhoa2016/qwen3-reranker-4b-hardened-llama-cpp-runtime`; chỉ khi exact path này không tồn tại mới dùng fail-closed recursive lookup bên trong `/kaggle/input/datasets`. Cell verify đủ tám entries trong `RUNTIME-MANIFEST.sha256`, copy bundle byte-identical sang `/kaggle/working`, set `LD_LIBRARY_PATH` và `LLAMA_SERVER_BIN`, verify launcher linkage và bắt buộc `LLAMA_CPP_SETUP=PASS_EXISTING_BINARY`. Generic runtime download/build fallback bị từ chối.


````python
import hashlib
import shutil
import traceback

print("RUNTIME_PREFLIGHT_BEGIN")

def _print_block(label, text):
    text = text or ""
    print(f"--- {label} BEGIN ---")
    print(text.rstrip() if text.strip() else "<empty>")
    print(f"--- {label} END ---")

def _run_logged(label, argv, *, cwd=None, env=None, timeout=None):
    print(f"{label}_COMMAND =", " ".join(map(str, argv)))
    result = subprocess.run(
        list(map(str, argv)),
        cwd=cwd,
        env=env,
        text=True,
        capture_output=True,
        check=False,
        timeout=timeout,
    )
    print(f"{label}_RETURN_CODE =", result.returncode)
    _print_block(f"{label}_STDOUT", result.stdout)
    _print_block(f"{label}_STDERR", result.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"{label} failed with return code {result.returncode}")
    print(f"{label}=PASS")
    return result

try:
    kaggle_input = Path("/kaggle/input")
    assert kaggle_input.is_dir(), "/kaggle/input is missing"
    top_level_mounts = sorted(p.name for p in kaggle_input.iterdir() if p.is_dir())
    print("KAGGLE_INPUT_TOP_LEVEL_COUNT =", len(top_level_mounts))
    print("KAGGLE_INPUT_TOP_LEVEL =", top_level_mounts)

    _run_logged(
        "PIP_EDITABLE_INSTALL",
        [sys.executable, "-m", "pip", "install", "-q", "-e", str(APP_ROOT)],
    )

    qdrant_setup = _run_logged(
        "QDRANT_SETUP",
        ["bash", str(APP_ROOT / "scripts/setup-qdrant-production-demo.sh")],
    )
    for line in qdrant_setup.stdout.splitlines():
        if line.startswith("QDRANT_BIN="):
            os.environ["QDRANT_BIN"] = line.split("=", 1)[1]
    assert os.environ.get("QDRANT_BIN"), "QDRANT_BIN was not produced by setup."
    print("QDRANT_BIN =", os.environ["QDRANT_BIN"])

    EXPECTED_LAUNCHER_SHA = "28a79707376877f09065fa05fda5a9a6f57dfb4aed01c9918123667e38ae1f41"
    EXPECTED_IMPL_SHA = "c4807f2f10cdf354270ac97c1f091d0846e8154749b4b8347f5f26a40184d425"
    EXPECTED_DATASET_OWNER = "dangkhoa2016"
    EXPECTED_DATASET_DIRNAME = "qwen3-reranker-4b-hardened-llama-cpp-runtime"

    datasets_root = kaggle_input / "datasets"
    print("KAGGLE_DATASETS_ROOT =", datasets_root)
    print("KAGGLE_DATASETS_ROOT_EXISTS =", datasets_root.is_dir())
    assert datasets_root.is_dir(), (
        "Expected Kaggle datasets namespace at /kaggle/input/datasets. "
        f"Top-level mounts were: {top_level_mounts}"
    )

    preferred_runtime_input = (
        datasets_root / EXPECTED_DATASET_OWNER / EXPECTED_DATASET_DIRNAME
    )
    print("RUNTIME_PREFERRED_PATH =", preferred_runtime_input)
    print("RUNTIME_PREFERRED_PATH_EXISTS =", preferred_runtime_input.is_dir())

    if (
        preferred_runtime_input.is_dir()
        and (preferred_runtime_input / "RUNTIME-MANIFEST.sha256").is_file()
    ):
        mount_candidates = [preferred_runtime_input]
        print("RUNTIME_LOCATOR_MODE=EXACT_NAMESPACED_PATH")
    else:
        mount_candidates = sorted(
            {
                manifest.parent.resolve()
                for manifest in datasets_root.rglob("RUNTIME-MANIFEST.sha256")
                if manifest.parent.name == EXPECTED_DATASET_DIRNAME
            },
            key=lambda p: str(p),
        )
        print("RUNTIME_LOCATOR_MODE=RECURSIVE_DATASETS_NAMESPACE")

    print("RUNTIME_MOUNT_CANDIDATES =", [str(p) for p in mount_candidates])
    assert len(mount_candidates) == 1, (
        "Expected exactly one attached canonical runtime dataset under "
        "/kaggle/input/datasets; "
        f"found {mount_candidates}. Attach Kaggle Dataset "
        "dangkhoa2016/qwen3-reranker-4b-hardened-llama-cpp-runtime."
    )
    runtime_input = mount_candidates[0]
    print("RUNTIME_MOUNT=PASS")
    print("RUNTIME_INPUT =", runtime_input)
    print("RUNTIME_INPUT_FILES =", [(p.name,p.stat().st_size) for p in sorted(runtime_input.iterdir()) if p.is_file()])

    manifest_path = runtime_input / "RUNTIME-MANIFEST.sha256"
    print("RUNTIME_MANIFEST_PATH =", manifest_path)
    assert manifest_path.is_file(), f"Missing {manifest_path}"

    manifest = {}
    rows = 0
    for raw in manifest_path.read_text().splitlines():
        raw = raw.strip()
        if not raw: continue
        rows += 1
        sha, name = raw.split(None, 1)
        name = name.strip()
        src = runtime_input / name
        exists = src.is_file()
        actual = hashlib.sha256(src.read_bytes()).hexdigest() if exists else None
        ok = exists and actual == sha
        print("RUNTIME_MANIFEST_ENTRY", name, f"exists={exists}", f"expected={sha}", f"actual={actual}", f"pass={ok}")
        assert exists, f"Runtime manifest file missing: {name}"
        assert actual == sha, f"Runtime input SHA mismatch: {name} {actual} != {sha}"
        manifest[name] = sha

    print(f"RUNTIME_MANIFEST={len(manifest)}/{rows} PASS")
    assert len(manifest) == 8, f"Expected exactly 8 runtime entries, found {len(manifest)}"
    assert manifest.get("llama-server-patched") == EXPECTED_LAUNCHER_SHA
    assert manifest.get("libllama-server-impl.so") == EXPECTED_IMPL_SHA
    print("RUNTIME_FROZEN_IDENTITIES=PASS")

    runtime_work = Path("/kaggle/working/qwen3-reranker-4b-hardened-llama-cpp-runtime")
    if runtime_work.exists(): shutil.rmtree(runtime_work)
    runtime_work.mkdir(parents=True)
    for name in manifest: shutil.copyfile(runtime_input / name, runtime_work / name)

    for name, sha in manifest.items():
        copied_path = runtime_work / name
        actual = hashlib.sha256(copied_path.read_bytes()).hexdigest()
        ok = actual == sha
        print("RUNTIME_COPY_ENTRY", name, f"expected={sha}", f"actual={actual}", f"pass={ok}")
        assert ok, f"Runtime copied SHA mismatch: {name} {actual} != {sha}"
    print(f"RUNTIME_COPY={len(manifest)}/{len(manifest)} PASS")

    launcher = runtime_work / "llama-server-patched"
    launcher.chmod(0o755)
    os.environ["LD_LIBRARY_PATH"] = str(runtime_work) + (":" + os.environ["LD_LIBRARY_PATH"] if os.environ.get("LD_LIBRARY_PATH") else "")
    os.environ["LLAMA_SERVER_BIN"] = str(launcher)
    print("RUNTIME_WORK =", runtime_work)
    print("LLAMA_SERVER_BIN =", os.environ["LLAMA_SERVER_BIN"])
    print("LD_LIBRARY_PATH =", os.environ["LD_LIBRARY_PATH"])

    _run_logged("LLAMA_LINKAGE_PROBE", [str(launcher), "--version"], env=os.environ.copy(), timeout=30)
    llama_setup = _run_logged("LLAMA_SETUP", ["bash", str(APP_ROOT / "scripts/setup_llama_cpp_b10699.sh")], env=os.environ.copy())
    assert "LLAMA_CPP_SETUP=PASS_EXISTING_BINARY" in llama_setup.stdout, (
        "Canonical reproduction must use the attached hardened runtime; download/build fallback is not allowed."
    )
    print("LLAMA_SETUP_MODE=PASS_EXISTING_BINARY")
    print("QUALIFIED_LLAMA_RUNTIME_INPUT=PASS")
    print("RUNTIME_PREFLIGHT=PASS")
except Exception as exc:
    print("RUNTIME_PREFLIGHT=FAIL")
    print("RUNTIME_PREFLIGHT_EXCEPTION_TYPE =", type(exc).__name__)
    print("RUNTIME_PREFLIGHT_EXCEPTION_MESSAGE =", str(exc))
    traceback.print_exc()
    raise
````

````python
import secrets
import urllib.request

run_root = Path(os.environ["PRODUCTION_DEMO_RUN_ROOT"])
run_root.mkdir(parents=True, exist_ok=True)

os.environ["DUAL_API_KEY"] = secrets.token_hex(24)
os.environ.update({
    "RERANKER_BACKEND": "llama_cpp",
    "MODEL_DTYPE": "float16",
    "QUANTIZATION_MODE": "none",
    "SERVER_HOST": "127.0.0.1",
    "SERVER_PORT": "8000",
    "LLAMA_SERVER_PORT": "8081",
    "LLAMA_SERVER_CONTEXT_SIZE": "1024",
    "MAX_RERANK_DOCUMENTS": str(RERANK_TOP_K),
    "HF_HUB_OFFLINE": "1",
    "TRANSFORMERS_OFFLINE": "1",
    "ALLOW_REMOTE_MODEL_DOWNLOAD": "0",
})

log_handle = open(run_root / "hybrid-server.log", "w")
hybrid = subprocess.Popen(
    ["bash", str(APP_ROOT / "scripts/start-server.sh")],
    stdout=log_handle,
    stderr=subprocess.STDOUT,
    env=os.environ.copy(),
)

(run_root / "hybrid-server.pid").write_text(f"{hybrid.pid}\n")

for _ in range(900):
    try:
        with urllib.request.urlopen(
            "http://127.0.0.1:8000/ready", timeout=3
        ) as response:
            if response.status == 200:
                break
    except Exception:
        pass

    if hybrid.poll() is not None:
        log_handle.flush()
        raise RuntimeError(
            (run_root / "hybrid-server.log").read_text(errors="replace")[-4000:]
        )
    time.sleep(1)
else:
    raise RuntimeError("hybrid readiness timeout")

print("HYBRID_READY=PASS pid=", hybrid.pid)
````

## 8. Prove the executing hardened llama runtime / Chứng minh hardened llama runtime đang thực thi

**English:** This cell proves the runtime actually executing. It finds the unique llama process serving the attached GGUF on port `8081`, hashes `/proc/<pid>/exe` and requires the frozen launcher SHA-256 `28a79707376877f09065fa05fda5a9a6f57dfb4aed01c9918123667e38ae1f41`, then parses `/proc/<pid>/maps` to resolve the mapped `libllama-server-impl.so` and requires implementation SHA-256 `c4807f2f10cdf354270ac97c1f091d0846e8154749b4b8347f5f26a40184d425`.

**Tiếng Việt:** Cell này chứng minh runtime thực sự đang chạy. Nó tìm unique llama process serve attached GGUF ở port `8081`, hash `/proc/<pid>/exe` và yêu cầu frozen launcher SHA-256 `28a79707376877f09065fa05fda5a9a6f57dfb4aed01c9918123667e38ae1f41`, sau đó parse `/proc/<pid>/maps` để resolve `libllama-server-impl.so` đang được map và yêu cầu implementation SHA-256 `c4807f2f10cdf354270ac97c1f091d0846e8154749b4b8347f5f26a40184d425`.


````python
EXPECTED_LAUNCHER_SHA = "28a79707376877f09065fa05fda5a9a6f57dfb4aed01c9918123667e38ae1f41"
EXPECTED_IMPL_SHA = "c4807f2f10cdf354270ac97c1f091d0846e8154749b4b8347f5f26a40184d425"

reranker_gguf = os.environ["RERANKER_GGUF_PATH"]
matches = []
for proc_dir in Path("/proc").iterdir():
    if not proc_dir.name.isdigit():
        continue
    try:
        cmdline = (proc_dir / "cmdline").read_bytes().replace(b"\0", b" ").decode("utf-8", "replace")
    except (OSError, PermissionError):
        continue
    if reranker_gguf in cmdline and ("--port 8081" in cmdline or "--port=8081" in cmdline):
        matches.append((int(proc_dir.name), cmdline))

assert len(matches) == 1, f"Expected one llama process, found: {matches}"
llama_pid, llama_cmdline = matches[0]

exe_real = Path(os.readlink(f"/proc/{llama_pid}/exe"))
exe_sha = hashlib.sha256(exe_real.read_bytes()).hexdigest()
assert exe_sha == EXPECTED_LAUNCHER_SHA, (
    f"Executing launcher SHA mismatch: pid={llama_pid} exe={exe_real} sha256={exe_sha}"
)

maps_lines = Path(f"/proc/{llama_pid}/maps").read_text(errors="replace").splitlines()
impl_paths = []
for line in maps_lines:
    if "libllama-server-impl.so" not in line:
        continue
    candidate = line.split()[-1]
    if candidate.startswith("/"):
        p = Path(candidate)
        if p.is_file():
            impl_paths.append(p.resolve())

impl_paths = sorted(set(impl_paths))
assert len(impl_paths) == 1, f"Expected one mapped implementation file, found: {impl_paths}"
impl_real = impl_paths[0]
impl_sha = hashlib.sha256(impl_real.read_bytes()).hexdigest()
assert impl_sha == EXPECTED_IMPL_SHA, (
    f"Mapped implementation SHA mismatch: path={impl_real} sha256={impl_sha}"
)

print("LLAMA_RUNTIME_PID =", llama_pid)
print("LLAMA_RUNTIME_EXE =", exe_real)
print("LLAMA_LAUNCHER_SHA256 =", exe_sha)
print("LLAMA_MAPPED_IMPLEMENTATION =", impl_real)
print("LLAMA_IMPLEMENTATION_SHA256 =", impl_sha)
print("LLAMA_EXECUTING_PROCESS_AND_MAPPED_IMPLEMENTATION=PASS")

(run_root / "llama-server.pid").write_text(f"{llama_pid}\n")
qdrant_pid = int((run_root / "qdrant/qdrant.pid").read_text().strip())
owner_record = {
    "source_head": SOURCE_HEAD,
    "run_root": str(run_root),
    "qdrant_pid": qdrant_pid,
    "hybrid_pid": hybrid.pid,
    "llama_pid": llama_pid,
    "ports": {"qdrant": 6333, "api": 8000, "reranker": 8081},
    "Qwen3-Reranker-4B_GGUF": Path(reranker_gguf).name,
    "llama_launcher_sha256": exe_sha,
    "llama_implementation_sha256": impl_sha,
}
(run_root / "qualification-owner.json").write_text(
    json.dumps(owner_record, indent=2) + "\n"
)
print("PROCESS_OWNERSHIP_RECORD=PASS")
````
